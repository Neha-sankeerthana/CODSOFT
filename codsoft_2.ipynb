```python
PLAYER_X = 'X'
PLAYER_O = 'O'
EMPTY_CELL = ' '

def display_board(board):
    """Prints the current state of the game board."""
    for row in board:
        print(" | ".join(row))
        print("---" * 3)

def check_winner(board, player):
    """Checks if the specified player has won."""
    for i in range(3):
        if all(board[i][j] == player for j in range(3)) or \
           all(board[j][i] == player for j in range(3)):
            return True
    if all(board[i][i] == player for i in range(3)) or \
       all(board[i][2 - i] == player for i in range(3)):
        return True
    return False

def board_is_full(board):
    """Returns True if the board is full, otherwise False."""
    return all(board[i][j] != EMPTY_CELL for i in range(3) for j in range(3))

def available_moves(board):
    """Returns a list of empty cells on the board."""
    return [(i, j) for i in range(3) for j in range(3) if board[i][j] == EMPTY_CELL]

def minimax(board, depth, is_maximizing):
    """Implements the Minimax algorithm to determine the best move."""
    if check_winner(board, PLAYER_X):
        return -1
    elif check_winner(board, PLAYER_O):
        return 1
    elif board_is_full(board):
        return 0

    if is_maximizing:
        max_score = float('-inf')
        for i, j in available_moves(board):
            board[i][j] = PLAYER_O
            score = minimax(board, depth + 1, False)
            board[i][j] = EMPTY_CELL
            max_score = max(max_score, score)
        return max_score
    else:
        min_score = float('inf')
        for i, j in available_moves(board):
            board[i][j] = PLAYER_X
            score = minimax(board, depth + 1, True)
            board[i][j] = EMPTY_CELL
            min_score = min(min_score, score)
        return min_score

def determine_best_move(board):
    """Finds the best move for the AI player using Minimax."""
    best_score = float('-inf')
    best_move = (-1, -1)

    for i, j in available_moves(board):
        board[i][j] = PLAYER_O
        move_score = minimax(board, 0, False)
        board[i][j] = EMPTY_CELL

        if move_score > best_score:
            best_move = (i, j)
            best_score = move_score

    return best_move

def start_game():
    """Main function to start the Tic Tac Toe game."""
    game_board = [[EMPTY_CELL] * 3 for _ in range(3)]

    while True:
        display_board(game_board)

        # Human player's turn
        try:
            row = int(input("Choose your row (0, 1, or 2): "))
            col = int(input("Choose your column (0, 1, or 2): "))

            if game_board[row][col] == EMPTY_CELL:
                game_board[row][col] = PLAYER_X
            else:
                print("That cell is already taken. Please try again.")
                continue
        except (ValueError, IndexError):
            print("Invalid input. Please enter numbers between 0 and 2.")
            continue

        if check_winner(game_board, PLAYER_X):
            display_board(game_board)
            print("Congratulations! You win!")
            break
        elif board_is_full(game_board):
            display_board(game_board)
            print("It's a draw!")
            break

        # AI's turn
        print("AI is making its move...")
        best_move = determine_best_move(game_board)
        game_board[best_move[0]][best_move[1]] = PLAYER_O

        if check_winner(game_board, PLAYER_O):
            display_board(game_board)
            print("AI wins! Better luck next time.")
            break
        elif board_is_full(game_board):
            display_board(game_board)
            print("It's a draw!")
            break

if __name__ == "__main__":
    start_game()
```
